````python
import re
import csv
import json
import time
import random
import openai
from openai import OpenAI
from openai import RateLimitError, APIError

# Set up OpenAI API client
client = OpenAI(api_key="***REMOVED-OPENAI-KEY***")  # <-- Replace with your key

INSTRUCTIONS = """\
Instructions for Grading:
For each prompt, you will see two clarification questions (Prompt A and Prompt B).
One of these comes from the fine‑tuned model, which has been optimized for clearer, more focused, and immediately actionable questions. As you rate, pay special attention to those qualities.

Please rate each question on the following criteria, keeping in mind that the fine‑tuned model is designed to:
  • Pinpoint missing details with precision  
  • Offer concrete examples or guidance you can act on instantly (e.g., “For example, should this function take a list of ints or strings?”)  
  • Align perfectly with the original task  

Precision and Focus: How exactly the clarification question zeroes in on the missing detail or ambiguity in the original prompt.
  1–3 (Generic): “Can you clarify?” doesn’t hint at what to clarify.  
  4 (Somewhat): Suggests a possible gap but still vague.  
  5–7 (Highly Focused): Pinpoints exactly what detail is missing so you know what to add.  
  *Bonus*: Use of “For example, …” with concrete options is a plus.

Immediate Editability: How easily you can rewrite your original prompt using only this question.
  1–3 (Still Guessing): You’d need more guidance.  
  4 (Partial Guidance): Gives some direction but leaves you wondering about specifics.  
  5–7 (Fully Actionable): You could immediately improve your prompt without further help.  
  *Plus*: Questions that provide example choices or templates score higher.

Contextual Fit: How well the question aligns with the intent of the original programming task.
  1–3 (Off‑Target): Asks about irrelevant aspects.  
  4 (Passable): Somewhat on topic but not ideal.  
  5–7 (Perfectly Aligned): Feels like the “right” question to ask, given the task.

Overall Preference: Which question do you prefer overall (A or B)?  
Focus on the question that best demonstrates the fine‑tuned model’s strengths in clarity, actionable guidance, precise relevance, and the use of concrete “For example, …” options.

NOTE: Both fine‑tuned and non‑fine‑tuned models were capped at 50 tokens. Please only grade the core clarification question text, ignoring any trailing artifacts or non‑English characters.
"""


def build_gpt_prompt(row):
    prompt = f"""{INSTRUCTIONS}

Original Prompt:
{row['Original Prompt']}

Prompt A:
{row['Prompt A']}

Prompt B:
{row['Prompt B']}

Please fill in the following for each prompt:
- Precision and Focus (A, 1-7):
- Precision and Focus (B, 1-7):
- Immediate Editability (A, 1-7):
- Immediate Editability (B, 1-7):
- Contextual Fit (A, 1-7):
- Contextual Fit (B, 1-7):
- Overall Preference (A or B):
- Free-Form Feedback:
"""
    return prompt

def call_gpt(prompt, model="gpt-4o", max_retries=3, temperature=0.0):
    for attempt in range(max_retries):
        try:
            response = client.chat.completions.create(
                model=model,
                messages=[{"role": "user", "content": prompt}],
                temperature=temperature,
                max_tokens=512
            )
            return response.choices[0].message.content.strip()
        except (RateLimitError, APIError) as e:
            wait = 5 + random.uniform(0, 5)
            print(f"Rate limit or API error: {e}. Retrying in {wait:.1f} seconds...")
            time.sleep(wait)
        except Exception as e:
            print(f"Unexpected error: {e}")
            break
    return None

def parse_gpt_response(gpt_response):
    result = {
        "Precision and Focus (A)": None,
        "Precision and Focus (B)": None,
        "Immediate Editability (A)": None,
        "Immediate Editability (B)": None,
        "Contextual Fit (A)": None,
        "Contextual Fit (B)": None,
        "Overall Preference": None,
        "Free-Form Feedback": None,
        "raw_gpt_response": gpt_response
    }
    if not gpt_response:
        return result

    # Patterns for each metric
    patterns = {
        "Precision and Focus (A)": r"Precision and Focus \(A[^\)]*\):\s*([0-9]+)",
        "Precision and Focus (B)": r"Precision and Focus \(B[^\)]*\):\s*([0-9]+)",
        "Immediate Editability (A)": r"Immediate Editability \(A[^\)]*\):\s*([0-9]+)",
        "Immediate Editability (B)": r"Immediate Editability \(B[^\)]*\):\s*([0-9]+)",
        "Contextual Fit (A)": r"Contextual Fit \(A[^\)]*\):\s*([0-9]+)",
        "Contextual Fit (B)": r"Contextual Fit \(B[^\)]*\):\s*([0-9]+)",
        "Overall Preference": r"Overall Preference \(A or B\):\s*([AB])",
        "Free-Form Feedback": r"Free-Form Feedback:\s*(.*)"
    }

    for key, pattern in patterns.items():
        match = re.search(pattern, gpt_response, re.DOTALL)
        if match:
            value = match.group(1).strip()
            if key.startswith("Precision") or key.startswith("Immediate") or key.startswith("Contextual"):
                try:
                    value = int(value)
                except ValueError:
                    pass
            result[key] = value

    # Free-Form Feedback may be multi-line; get everything after it
    if "Free-Form Feedback:" in gpt_response:
        feedback = gpt_response.split("Free-Form Feedback:", 1)[-1].strip()
        result["Free-Form Feedback"] = feedback

    return result

def main():
    input_csv = "/Users/harshdarji/Downloads/responses_for_participants_anonymized.csv"
    output_json = "/Users/harshdarji/Downloads/gpt4o_grading_results.json"
    results = []

    with open(input_csv, newline='', encoding='utf-8') as f:
        reader = csv.reader(f)
        header = None
        for row in reader:
            # Find the header row (the one containing 'Prompt ID')
            if not header:
                if "Prompt ID" in row:
                    header = row
                continue
            # Map row to dict
            row_dict = dict(zip(header, row))
            # Skip empty rows
            if not row_dict.get("Prompt ID") or not row_dict.get("Original Prompt"):
                continue
            # Skip rows where Prompt A or B is missing
            if not row_dict.get("Prompt A") or not row_dict.get("Prompt B"):
                continue
            gpt_prompt = build_gpt_prompt(row_dict)
            print(f"Grading Prompt ID {row_dict['Prompt ID']} ...")
            gpt_response = call_gpt(gpt_prompt)
            parsed = parse_gpt_response(gpt_response)
            results.append({
                "Prompt ID": row_dict["Prompt ID"],
                "Original Prompt": row_dict["Original Prompt"],
                "Prompt A": row_dict["Prompt A"],
                "Prompt B": row_dict["Prompt B"],
                **parsed
            })
            # Optional: save progress every 10
            if len(results) % 10 == 0:
                with open(output_json, "w", encoding="utf-8") as out:
                    json.dump(results, out, ensure_ascii=False, indent=2)

    with open(output_json, "w", encoding="utf-8") as out:
        json.dump(results, out, ensure_ascii=False, indent=2)
    print(f"Saved {len(results)} graded prompts to {output_json}")

if __name__ == "__main__":
    main()
````

````python
import json

# Load the grading results and the original responses
with open('/Users/harshdarji/Downloads/gpt4o_grading_results.json', 'r', encoding='utf-8') as f:
    graded = json.load(f)
with open('/Users/harshdarji/Downloads/cleaned_responses.json', 'r', encoding='utf-8') as f:
    original = json.load(f)

# Build a lookup from prompt to raw/fine-tuned response
orig_lookup = {}
for item in original:
    orig_lookup[item['prompt'].strip()] = {
        "raw": item['raw-response'].strip(),
        "finetuned": item['finetuned-response'].strip()
    }

# Now, for each graded entry, figure out which is which
for entry in graded:
    prompt = entry['Original Prompt'].strip()
    prompt_a = entry['Prompt A'].strip()
    prompt_b = entry['Prompt B'].strip()
    mapping = {}
    responses = orig_lookup.get(prompt, {})
    if not responses:
        print(f"Prompt not found: {prompt[:40]}...")
        continue
    # Use substring matching (strip possible whitespace, newlines, etc.)
    if prompt_a[:50] in responses['raw'] or responses['raw'][:50] in prompt_a:
        mapping['A'] = 'raw'
        mapping['B'] = 'finetuned'
    elif prompt_a[:50] in responses['finetuned'] or responses['finetuned'][:50] in prompt_a:
        mapping['A'] = 'finetuned'
        mapping['B'] = 'raw'
    else:
        # fallback: use more lenient matching
        if prompt_a in responses['raw']:
            mapping['A'] = 'raw'
            mapping['B'] = 'finetuned'
        elif prompt_a in responses['finetuned']:
            mapping['A'] = 'finetuned'
            mapping['B'] = 'raw'
        else:
            print(f"Could not match for prompt: {prompt[:40]}...")
            mapping['A'] = mapping['B'] = 'unknown'
    entry['A_model'] = mapping['A']
    entry['B_model'] = mapping['B']

# Now you can aggregate by model
metrics = [
    "Precision and Focus",
    "Immediate Editability",
    "Contextual Fit"
]
scores = {m: {"raw": [], "finetuned": []} for m in metrics}
preferences = {"raw": 0, "finetuned": 0}

for entry in graded:
    for ab in ["A", "B"]:
        model = entry.get(f"{ab}_model")
        if model not in ["raw", "finetuned"]:
            continue
        for m in metrics:
            key = f"{m} ({ab})"
            val = entry.get(key, "")
            try:
                val = int(val)
            except Exception:
                continue
            scores[m][model].append(val)
    # Preference
    pref = entry.get("Overall Preference", "").strip().upper()
    if pref in ["A", "B"]:
        model = entry.get(f"{pref}_model")
        if model in preferences:
            preferences[model] += 1

# Print results
print("=== Average Scores by Metric and Model ===")
for m in metrics:
    for model in ["raw", "finetuned"]:
        vals = scores[m][model]
        avg = sum(vals)/len(vals) if vals else 0
        print(f"{m} ({model}): {avg:.2f} (n={len(vals)})")
    print()

total = sum(preferences.values())
print("=== Overall Preference ===")
for model in ["raw", "finetuned"]:
    pct = 100 * preferences[model] / total if total else 0
    print(f"{model}: {preferences[model]} times ({pct:.1f}%)")
print(f"Total graded: {total}")
````
